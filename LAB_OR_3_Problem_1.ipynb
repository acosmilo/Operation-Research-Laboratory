# Code for Expand Model

```python
import ortools
print("OR-Tools se ha instalado correctamente.")
```

```python
from ortools.linear_solver import pywraplp
solver = pywraplp.Solver.CreateSolver('GLOP')
```

```python
x1 = solver.NumVar(0, solver.infinity(), 'x1')
x2 = solver.NumVar(0, solver.infinity(), 'x2')
print("Numero de variables =", solver.NumVariables())
# 0 significa que no puede tomar valores menores a cero
# infinity significa que no tiene límite superior
```

```python
solver.Add(x1 + x2  <= 10)
solver.Add(4*x1 + 2*x2 <= 12)
solver.Add(2*x1 + 2*x2  <= 9)
print("Numero de restricciones =", solver.NumConstraints())
```

```python
solver.Maximize(3* x1 + 2.5 * x2 )
```

```python
status = solver.Solve()
```

```python
if status == pywraplp.Solver.OPTIMAL:
        print('Solución óptima Z:')
        print('Valor de x1:', x1.solution_value())
        print('Valor de x2:', x2.solution_value())
        print('Valor óptimo de la función objetivo (Z):', solver.Objective().Value())
else:
        print('No se encontró una solución óptima.')
```
